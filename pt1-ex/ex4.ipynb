```python
data = [3.67 , 1.82, 3.73, 4.10, 4.30, 1.28, 8.14, 2.43, 4.17, 5.36, 3.96, 6.54, 5.84, 7.35, 3.63, 2.93, 2.82, 8.45, 5.28, 5.41, 7.77, 4.65, 1.88, 2.12, 4.26, 2.78, 5.54, 0.90, 5.09, 4.07]
```

```python
import pandas as pd 
import matplotlib.pyplot as plt

dt = pd.DataFrame(data, columns=['Distribuição de Valores'])

dt.value_counts()
dt.hist(bins=6, color='pink')
plt.show
```

```python
import numpy as np

df = pd.DataFrame(data, columns=['valores'])
df['y'] = np.random.normal(0, 0.02, size=len(data))

fig, ax = plt.subplots(figsize=(8,3))
ax.scatter(df['valores'], df['y'], alpha=0.7, color='pink')

ax.set_yticks([])
ax.set_xlabel('Valores')
ax.set_title('Gráfico de Dispersão Unidimensional')

plt.show()
```
