```python
import pandas as pd 

data = pd.read_csv('../data/CD-Brasil.csv')
print(data.head())
```

## Distribuição de Frequências da População e Densidade populacional

```python
# frequencia absoluta população 
fi_pop = data['populacao'].value_counts()

#frequencia relativa população
ni_pop = data['populacao'].value_counts(normalize=True) * 100

#frequencia acumulada população
fa_pop = fi_pop.cumsum()
```

```python
# frequencia absoluta densidade 
fi_dens = data['densidade_hab_km2'].value_counts()

#frequencia relativa densidade
ni_dens = data['densidade_hab_km2'].value_counts(normalize=True) * 100

#frequencia acumulada densidade
fa_dens = fi_pop.cumsum()
```

```python
tabela_geral = pd.DataFrame({
    ('População', 'Classe'): fi_pop.index,
    ('População', 'Absoluta (fi)'): fi_pop.values,
    ('População', 'Relativa (ni)'): ni_pop.values,
    ('População', 'Acumulada (Fi)'): fa_pop.values,
    
    ('Densidade', 'Classe'): fi_dens.index,
    ('Densidade', 'Absoluta (fi)'): fi_dens.values,
    ('Densidade', 'Relativa (ni)'): ni_dens.values,
    ('Densidade', 'Acumulada (Fi)'): fa_dens.values
})

print(tabela_geral)
```
