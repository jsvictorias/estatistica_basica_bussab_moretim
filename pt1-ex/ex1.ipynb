```python
import pandas as pd

data = pd.read_excel("../data/data_one.xlsx", header=1)
print(data.head())
```

## Distribuição de Frequências Para Variáveis Discretas

```python
# frequencia absoluta 
fi = data['Estado Civil'].value_counts()

# frequencia relativa
ni = data['Estado Civil'].value_counts(normalize=True) * 100
```

```python
tabela = pd.DataFrame({
    'Frequência Absoluta (f_i)': fi,
    'Frequência Relativa (%)': ni.round(1),
    'Frequência Acumulada': fi.cumsum()
})
```

```python
print(tabela)
```

```python
# frequencia absoluta
fi_rp = data['Região de Procedência'].value_counts()

# frquencia relativa
ni_rp = data['Região de Procedência'].value_counts(normalize=True) * 100

tabela_rp = pd.DataFrame({
    'Frequência Absoluta': fi_rp,
    'Frequência Relativa (%)': ni_rp.round(1),
    'Frequência Acumulada': fi_rp.cumsum()
})

print(tabela_rp)
```

## Distribuição de Frequências Para Variáveis Contínuas

```python
# Criando intervalo para os dados 
data['Anos'] = pd.cut(data['Anos'], bins=4)

# frequencia absoluta 
fi_idade = data['Anos'].value_counts()

# frequencia relativa
ni_idade = data['Anos'].value_counts(normalize=True) * 100

tabela_idade = pd.DataFrame({
    'Frequência Absoluta': fi_idade,
    'Frequência Relativa (%)': ni_idade.round(1),
    'Frequência Acumulada': fi_idade.cumsum()
})
print(tabela_idade)
```
