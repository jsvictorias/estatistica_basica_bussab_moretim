```python
import matplotlib.pyplot as plt
import pandas as pd 

data = pd.read_excel('../data/data_one.xlsx', header=1)
print(data.head())
```

```python
import sys; print(sys.executable)
```

```python
mediana = data['Anos'].median()
print(mediana)
q3 = data['Anos'].quantile(0.75)
print(q3)
q1 = data['Anos'].quantile(0.25)
print(q1)

iiq = q3 - q1
print(iiq)


fens_inferior = q1 - (1.5 * iiq)
fens_superior = q3 + (1.5 * iiq)

print(f'fens superior: {fens_superior.round(1)}, fens inferior: {fens_inferior.round(1)}')
```

```python
df = pd.DataFrame(data)
df['Anos'].hist(bins=6)

plt.show()
```

```python
# frequencia 
contagem = data['Grau de Instrução'].value_counts()
contagem.plot.bar(color='pink', edgecolor='black')

plt.title('Distribuição Grau de Instrução')
plt.xlabel('Grau de Instrução')
plt.ylabel('Quantidade de Pessoas')

plt.show()
```
