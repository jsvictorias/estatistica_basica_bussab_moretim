```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('../data/CD-Municipios.csv')
print(data.head())
```

```python
print(data['populacao_10mil_hab'].describe().round(2))

plt.figure(figsize=(8,5))
plt.hist(data['populacao_10mil_hab'], bins='fd', edgecolor='black')
plt.xlabel('População (10 mil hab)')
plt.ylabel('Frequência (nº de municípios)')
plt.title('Distribuição da população dos municípios')
plt.show()
```
